## Parte 1 2 y 3

```python
import numpy as np
import matplotlib.pyplot as plt
T_values = [0.25, 0.5, 1, 2]
n = 252
num_simulations = 5000

results = {
    'final_values': {},
    'mean': {},
    'variance': {},
    'std': {}
}

for T in T_values:
    dt = T / n
    final_values = []
    
    for _ in range(num_simulations):
        increments = np.random.normal(0, np.sqrt(dt), size=n)
        final_values.append(np.sum(increments))
    
    results['final_values'][T] = final_values
    results['mean'][T] = np.mean(final_values)
    results['variance'][T] = np.var(final_values)
    results['std'][T] = np.std(final_values)

plt.figure(figsize=(12, 7))

colors = ["#2605FF", "#79803c", "#9C9C9C", "#ffffffff"]

for idx, T in enumerate(T_values):
    plt.hist(results['final_values'][T], bins=50, alpha=0.5, 
             label=f'T={T} (μ={results["mean"][T]:.3f}, σ={results["std"][T]:.3f})',
             color=colors[idx], edgecolor='black', linewidth=0.5)

plt.title('Distribución de W(T) para diferentes valores de T', fontsize=14, fontweight='bold')
plt.xlabel('W(T)', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)
plt.legend(loc='upper right', fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("=" * 60)
print("RESUMEN DE ESTADÍSTICAS")
print("=" * 60)
for T in T_values:
    print(f"\nT = {T}")
    print(f"  Media:              {results['mean'][T]:.6f}")
    print(f"  Desviación estándar: {results['std'][T]:.6f}")
    print(f"  Varianza:           {results['variance'][T]:.6f}")
print("=" * 60)
```
